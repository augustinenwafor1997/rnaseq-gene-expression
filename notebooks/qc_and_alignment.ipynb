```r
#Downloading FastQ files----

urls <- c(
  "ftp.sra.ebi.ac.uk/vol1/fastq/SRR592/006/SRR5922596/SRR5922596_1.fastq.gz",
  "ftp.sra.ebi.ac.uk/vol1/fastq/SRR592/006/SRR5922596/SRR5922596_2.fastq.gz",
  "ftp.sra.ebi.ac.uk/vol1/fastq/SRR592/006/SRR5922606/SRR5922606_1.fastq.gz",
  "ftp.sra.ebi.ac.uk/vol1/fastq/SRR592/006/SRR5922606/SRR5922606_2.fastq.gz"
)

for (url in urls) {
  filename <- basename(url) # Extract filename from URL
  cat(paste("Downloading", filename, "...\n"))
  # Use system() to run wget command. -O specifies output file name.
  # -c allows resuming partially downloaded files.
  # --tries=5 retries the download up to 5 times.
  # --timeout=60 sets a timeout for each connection attempt.
  system(paste0("wget -c --tries=5 --timeout=60 ", url, " -O ", filename))
  if (file.exists(filename)) {
    cat(paste(filename, "downloaded successfully.\n"))
  } else {
    cat(paste("Failed to download", filename, ". Please check the URL or network connection.\n"))
  }
}
```

# **Adapter Trimming**

```r
if (!requireNamespace("ShortRead", quietly = TRUE)) {
  BiocManager::install("ShortRead")
}
library(ShortRead)
```

```r
# Define adapter sequence
adapter_seq <- "AGATCGGAAGAGCACACGTCTGAACTCCAGTCA"

# Process both reads for SRR5922596
fq1_2596 <- readFastq("SRR5922596_1.fastq.gz")
fq2_2596 <- readFastq("SRR5922596_2.fastq.gz")

# Trim adapter from read 1 and read 2
trimmed_fq1_2596 <- trimLRPatterns(Rpattern = adapter_seq, subject = sread(fq1_2596))
trimmed_fq2_2596 <- trimLRPatterns(Rpattern = adapter_seq, subject = sread(fq2_2596))

# Keep only non-empty trimmed reads
non_empty_1 <- width(trimmed_fq1_2596) > 0
non_empty_2 <- width(trimmed_fq2_2596) > 0

# Subset and save
fq1_2596 <- fq1_2596[non_empty_1]
fq2_2596 <- fq2_2596[non_empty_2]

writeFastq(fq1_2596, "trimmed_SRR5922596_1.fastq.gz", compress=TRUE)
writeFastq(fq2_2596, "trimmed_SRR5922596_2.fastq.gz", compress=TRUE)
```

```r
#Building Index----
buildindex("Mm_full_index_rsubread",
           "Mus_musculus.NCBIM37.66.dna_rm.toplevel.fa",
           indexSplit=F)

reads1 <- list.files(pattern = "_1.fastq.gz$" )
reads2 <- list.files(pattern = "_2.fastq.gz$" )
all.equal(length(reads1),length(reads2))
```

```r
#Performing Alignment----
align(index="Mm_full_index_rsubread",
      readfile1=reads1,
      readfile2=reads2,
      input_format="gzFASTQ",
      output_format="BAM",
      nthreads=10)
```
